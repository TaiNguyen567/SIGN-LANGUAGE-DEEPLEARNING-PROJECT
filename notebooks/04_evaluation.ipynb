# Evaluation
Metrics and plots are produced only from a held-out test split and a trained checkpoint. The split should be signer-independent when signer IDs are available.

```python
from pathlib import Path
import json, subprocess, sys

project = Path('..').resolve()
checkpoint = project / 'checkpoints' / 'best_model.pt'
if checkpoint.is_file():
    subprocess.run([sys.executable, str(project / 'evaluate.py'), '--checkpoint', str(checkpoint), '--config', str(project / 'config.yaml')], cwd=project, check=True)
    report = project / 'reports' / 'metrics.json'
    display(json.loads(report.read_text(encoding='utf-8')))
else:
    print('No trained checkpoint exists; evaluation metrics are not available.')
```
