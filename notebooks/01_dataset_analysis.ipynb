# Dataset analysis
This notebook reads the project's UTF-8 transcript metadata. It does not create example data. Video/signer counts and transcript-length distributions are computed only when a real CSV is present.

```python
from pathlib import Path
import pandas as pd

metadata_path = Path('../dataset/metadata/all.csv')
if not metadata_path.is_file():
    print('No dataset supplied. See ../DATASET_SETUP.md; no statistics are inferred.')
else:
    data = pd.read_csv(metadata_path, keep_default_na=False)
    print(f'Videos: {len(data):,}')
    print(f'Signers: {data.subject_id.nunique() if "subject_id" in data else "not provided"}')
    data['word_count'] = data.text.astype(str).str.split().str.len()
    display(data[['word_count']].describe())
    display(data.language.value_counts(dropna=False) if 'language' in data else 'language column absent')
```
