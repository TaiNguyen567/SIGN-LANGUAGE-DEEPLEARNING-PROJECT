# Realtime test
The synthetic command validates model/decoder wiring with random inputs only. For real recognition, prepare data, train a checkpoint, and then run the continuous webcam command from a local terminal.

```python
from pathlib import Path
import subprocess, sys

project = Path('..').resolve()
subprocess.run([sys.executable, str(project / 'inference.py'), '--demo', '--device', 'auto'], cwd=project, check=True)
checkpoint = project / 'checkpoints' / 'best_model.pt'
if checkpoint.is_file():
    print('To open the local webcam, run: python inference.py --camera 0 --device auto')
else:
    print('No trained VSL model; the synthetic demo above is not recognition.')
```
