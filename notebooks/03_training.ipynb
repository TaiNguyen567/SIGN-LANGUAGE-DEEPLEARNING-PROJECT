# Training
Run the same CTC training entry point used by the command line. It writes epoch loss, CER, WER, and learning rate to `logs/training.csv`.

```python
from pathlib import Path
import subprocess, sys

project = Path('..').resolve()
train_csv = project / 'dataset' / 'metadata' / 'train.csv'
if train_csv.is_file():
    subprocess.run([sys.executable, str(project / 'train.py'), '--config', str(project / 'config.yaml'), '--device', 'auto'], cwd=project, check=True)
else:
    print('No training split found. Prepare a licensed dataset first.')
```
